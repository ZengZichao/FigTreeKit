# 01 — Basic Styling with FigTreeKit

This notebook demonstrates basic FigTreeKit usage:
- Loading a tree from Newick format
- Configuring layout and appearance
- Setting tip labels
- Exporting a styled Nexus file for FigTree

```python
from figtreekit import FigTreeStyler, LayoutType, TransformType
from pathlib import Path

# Ensure output directory exists
Path("output").mkdir(exist_ok=True)
```

## 1. Load a Tree

FigTreeKit can load trees from files or directly from strings.
Here we use the synthetic benchmark tree `mammalian_mitogenome.tre`,
whose tip labels are anonymous codes (M0001–M0030).

```python
# Load from file
styler = FigTreeStyler("../examples/benchmarks/mammalian_mitogenome.tre")
print(f"Loaded tree with {len(styler._tree_content.split(','))} taxa")
```

## 2. Configure Layout

Set the tree layout to polar for a circular visualization.

```python
styler.set_layout(LayoutType.POLAR)
styler.set_polar_layout(
    angular_range=360,
    root_angle=90,
    align_tip_labels=True
)
```

## 3. Style Tip Labels

Configure font name, size, and style for leaf labels.

```python
styler.set_tip_labels(
    is_shown=True,
    font_name="Arial",
    font_size=10,
    font_style=0  # 0=plain, 1=bold, 2=italic, 3=bold-italic
)
```

## 4. Configure Appearance

Set branch width and colors.

```python
styler.set_appearance(
    branch_line_width=1.5,
    foreground_color="#000000"
)
```

## 5. Export

Export the styled tree as a Nexus file. This file can be directly opened in FigTree.

```python
output_path = "output/01_basic_styled.nex"
styler.export(output_path)
print(f"Exported to: {output_path}")
print(f"File size: {Path(output_path).stat().st_size} bytes")
```

## 6. Method Chaining

FigTreeKit supports fluent API — chain all operations in one expression.

```python
FigTreeStyler("../examples/benchmarks/mammalian_mitogenome.tre") \
    .set_layout(LayoutType.RECTILINEAR) \
    .set_tip_labels(is_shown=True, font_name="Helvetica", font_size=8) \
    .set_appearance(branch_line_width=1.0) \
    .set_scale_bar(is_shown=True) \
    .export("output/01_rectilinear_styled.nex")

print("Exported rectilinear layout.")
```
