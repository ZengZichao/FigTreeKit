# 02 — Clade Highlighting

This notebook demonstrates how to highlight specific clades in a phylogenetic tree.

The example tree is fully synthetic: tip labels are anonymous codes (`T001`–`T010`),
and the highlighted groups are artificial clades used only for demonstration.

```python
from figtreekit import FigTreeStyler, LayoutType
from pathlib import Path

Path("output").mkdir(exist_ok=True)
```

## 1. Load a Synthetic Tree

We use a small artificial tree with anonymous tip codes.

```python
tree = (
    "((((T001:0.05,T002:0.05):0.10,(T003:0.07,T004:0.07):0.08):0.15,
       ((T005:0.06,T006:0.06):0.09,(T007:0.08,T008:0.08):0.07):0.16):0.20,
      (T009:0.12,T010:0.12):0.30);"
)
styler = FigTreeStyler().load_content(tree)
print("Tree loaded.")
```

## 2. Highlight Clades

Use `highlight_clade()` to add colored background highlights to groups of tips.

```python
group_a = ["T001", "T002", "T003", "T004"]
group_b = ["T005", "T006", "T007", "T008"]
group_c = ["T009", "T010"]

styler.highlight_clade(group_a, color="#FFCDD2", width=10, offset=0.5)
styler.set_clade_color_all(group_a, color="#C62828")

styler.highlight_clade(group_b, color="#BBDEFB", width=10, offset=0.5)
styler.set_clade_color_all(group_b, color="#1565C0")

styler.highlight_clade(group_c, color="#C8E6C9", width=10, offset=0.5)
styler.set_clade_color_all(group_c, color="#2E7D32")

print("Highlighted three synthetic clades.")
```

## 3. Configure Layout and Export

```python
styler.set_layout(LayoutType.RECTILINEAR)
styler.set_tip_labels(is_shown=True, font_name="Arial", font_size=10)
styler.set_appearance(branch_line_width=1.5)
styler.set_scale_bar(is_shown=True)

output_path = "output/02_clade_highlighted.nex"
styler.export(output_path)
print(f"Exported to: {output_path}")
```

## Summary

| Method | Purpose |
|---|---|
| `highlight_clade()` | Add colored background highlight to a clade |
| `set_clade_color_all()` | Change branch color for a clade |
| `set_clade_font()` | Set custom font for clade tip labels |
| `clear_annotations()` | Remove all clade annotations |
